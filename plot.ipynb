```python
import matplotlib.pyplot as plt
import numpy as np

# --- Data based on the paper's description ---
# The data points reflect the scenario described in the QuaSAR paper:
# - A significant jump in accuracy from 5% to 10% data.
# - Marginal gains from 10% to 20%, illustrating the plateau.
data_fractions = np.array([5, 10, 15, 20])
accuracies = np.array([71.1, 95.0, 95.6, 95.9])

# --- Plotting ---
# Set a style for a more professional look.
# 'seaborn-v0_8-paper' is a good choice for publications.
# Other options include 'seaborn-v0_8-whitegrid', 'ggplot', or 'default'.
plt.style.use('seaborn-v0_8-whitegrid')

# Create the figure and axes objects
fig, ax = plt.subplots(figsize=(8, 5))

# Plot the data
ax.plot(data_fractions, accuracies, marker='o', linestyle='-', color='b', label='QuaSAR Performance')

# --- Customization for Publication Quality ---
# Set the title and labels with appropriate font sizes
ax.set_title('Model Performance vs. Training Data Fraction', fontsize=16)
ax.set_xlabel('Training Data Fraction (%)', fontsize=12)
ax.set_ylabel('Test Accuracy (%)', fontsize=12)

# Set the limits for the axes to focus on the plateau
ax.set_xlim(0, 25)
ax.set_ylim(70, 100)

# Set the ticks for clarity
ax.set_xticks(data_fractions)
ax.set_yticks(np.arange(70, 101, 5))

# Add a grid for better readability
ax.grid(True, which='both', linestyle='--', linewidth=0.5)

# Add a legend
ax.legend(fontsize=12)

# Add text annotations for key points
ax.annotate('Significant improvement',
            xy=(10, 95.0),
            xytext=(7, 85),
            arrowprops=dict(facecolor='black', shrink=0.05),
            fontsize=12)

ax.annotate('Performance Plateau',
            xy=(17.5, 95.75),
            xytext=(15, 80),
            arrowprops=dict(facecolor='black', shrink=0.05),
            fontsize=12)


# --- Final Touches ---
# Ensure the layout is tight and there's no wasted space
plt.tight_layout()

# Save the figure in a high-resolution format suitable for publications (e.g., PDF, EPS, PNG)
plt.savefig('qusar_data_fraction_plateau.pdf', format='pdf', dpi=300)
plt.savefig('qusar_data_fraction_plateau.png', format='png', dpi=300)


# Show the plot
plt.show()
```
